### ⚡ Mathematical Formula

\[
E = P \times t
\]

Where:

- **E** = Electrical Energy (Wh)
- **P** = Power (W)
- **t** = Operating Time (hours)

Conversion:

\[
E(kWh) = \frac{P(W)\times t(h)}{1000}
\]

# ⚡ Task 04: AC Circuit Analysis Using Python Complex Numbers

### 🎯 Objective
To use Python complex numbers to calculate resistance, reactance, total impedance, magnitude, phase angle, and current in a series RLC circuit.

````python
import math
import cmath
````

````python
resistance = 20.0       # Resistance (Ohm)
inductance = 0.08       # Inductance (H)
capacitance = 100e-6    # Capacitance (F)
frequency = 50.0        # Frequency (Hz)
source_voltage = 230.0  # Source voltage (V)
````

````python
omega = 2 * math.pi * frequency

z_r = complex(resistance, 0)
z_l = complex(0, omega * inductance)
z_c = complex(0, -1 / (omega * capacitance))

z_total = z_r + z_l + z_c

current = source_voltage / z_total
````

````python
print(f"Total impedance: {z_total:.2f} Ω")
print(f"Resistance: {z_total.real:.2f} Ω")
print(f"Net reactance: {z_total.imag:.2f} Ω")
print(f"Impedance magnitude: {abs(z_total):.2f} Ω")
print(f"Phase angle: {math.degrees(cmath.phase(z_total)):.2f}°")
print(f"Current magnitude: {abs(current):.2f} A")
````

````python
if z_total.imag > 0:
    print("Circuit Type: Predominantly Inductive")
elif z_total.imag < 0:
    print("Circuit Type: Predominantly Capacitive")
else:
    print("Circuit Type: Purely Resistive")
````

### 📊 Result

The total circuit is predominantly **capacitive** because the imaginary part of the total impedance is negative.

- Inductive reactance: 25.13 Ω
- Capacitive reactance: 31.83 Ω
- Net reactance: -6.70 Ω

Since Xc > Xl, the circuit exhibits capacitive behavior.

### 💡 Conclusion

Python complex numbers make it easier to analyze AC circuits by representing resistance and reactance in a single impedance value. We successfully calculated the total impedance, phase angle, and current.
